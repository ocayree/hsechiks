```python
# В этом блоке мы парсим авито с целью получения средней цены куры-несушки
# Информация нужна для оценки функции издержек
# Используется selenium
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
import time
import numpy as np
# импорт библиотек
```

```python
# доп опции для корректного запуска хром
options = webdriver.ChromeOptions()
options.add_argument('--headless')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
driver = webdriver.Chrome(options=options)


actions = ActionChains(driver)
prices = [] #список для цен
names = [] # список для названий объявлений

url = "https://www.avito.ru/moskva_i_mo/ptitsy?cd=1&q=куры+несушки"
driver.get(url)

# динамический парсинг с небольшой задержкой в 5 сек
actions.send_keys(Keys.END).perform()
time.sleep(5)
actions.send_keys(Keys.END).perform()
time.sleep(5)
actions.send_keys(Keys.END).perform()
time.sleep(5)

# названия объявлений
names = driver.find_elements(By.CLASS_NAME, "iva-item-titleStep-zichc")
names = np.array([name.text for name in names])

# цены в объявлениях
prices = driver.find_elements(By.CSS_SELECTOR, "meta[itemprop='price']")
prices = np.array([float(price.get_attribute('content')) for price in prices])
```

```python

# сортировка
mask1 = prices != 0.
names, prices = names[mask1], prices[mask1]

mask2 = names != ""
df = pd.DataFrame(names[mask2], prices[mask2])
df.head()
```

```python
# искомая функция
def mean_chicken_price():
    return np.mean(prices)
```
