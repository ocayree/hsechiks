```python
import requests
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
import pandas as pd
from selenium.webdriver.common.action_chains import ActionChains
import time
```

```python
driver = webdriver.Chrome()
driver.get("https://5ka.ru/search/?value=яйца")
time.sleep(5)
actions = ActionChains(driver)
actions.scroll_by_amount(delta_x=0, delta_y=1000).perform()
time.sleep(5)
actions.scroll_by_amount(delta_x=0, delta_y=1000).perform() 
goods = driver.find_elements(By.CLASS_NAME, "chakra-stack.KnkuqE3h-.HmedBAdm-.css-8g8ihq")
```

```python
names = []
prices = []
for good in goods:
    name = good.find_element(By.CLASS_NAME, "chakra-text.SdLEFc2B-.css-1jdqp4k").text
    price_rub = int(good.find_element(By.CLASS_NAME, "chakra-text.DUXYWqnZ-.css-6uvdux").text)
    price_kop = int(good.find_element(By.CLASS_NAME, "chakra-text.j2bifgeA-.css-6uvdux").text)
    price = price_rub + price_kop / 100
    names.append(name)
    prices.append(price)
```

```python
raw_eggs_data_dict = dict(zip(names, prices))
```

```python
eggs_data_dict = {
    'Название': [],
    'Категория': [],
    'Цена': []
}

eggs_data = pd.DataFrame(eggs_data_dict)
```

```python
categories = ['СВ', 'СО', 'С1', 'С2']
for name in raw_eggs_data_dict:
    for category in categories:
        if category in name:
            new_row = pd.DataFrame({"Название" : [name], "Категория" : [category], "Цена" : [raw_eggs_data_dict[name]]})
            eggs_data = pd.concat([eggs_data, new_row], ignore_index = True)
            break
        
    
```
